```python
# Create color map based on continent
continent_colors = {
    'Asia': '#FF6B6B',
    'Europe': '#4ECDC4',
    'North America': '#45B7D1',
    'South America': '#96CEB4',
    'Africa': '#FECA57',
    'Oceania': '#FF9FF3',
    'Unknown': '#CAD3C8'
}
```

```python

# Fetch data
url = "https://disease.sh/v3/covid-19/countries"
data = requests.get(url).json()

target = 'deaths'  # Change to 'deaths', 'recovered', or 'active' as needed
N = 50  # Top N countries
sorting = True  # Whether to sort the final output by ratio
ratio = True

# Create DataFrame
df = pd.DataFrame(data)
```

```python
df['ratio']= df['deaths'] / df['cases'] * 100
#df['country'] = df['country'].apply(lambda x: x['name'] if isinstance(x, dict) else x)
#df = df[['country', 'cases', 'deaths', 'recovered', 'active', 'ratio', 'continent']]
# Add continent column using our comprehensive mapping
#df['continent'] = df['country'].map(country_continent_map).fillna('Unknown')
#df['continent'] = df['country'].fillna('Unknown')
df['color'] = df['continent'].map(continent_colors).fillna('#CAD3C8')
# df.info()
# df.head(1).countryInfo

for c,k in zip(df.head(10).country.items(),df.head(10).countryInfo.items()):
    print(c[1],k[1])
```

```python
if ratio:
    df_top15 = df.nlargest(N, 'ratio')
    format_str = '{:.2f}%'
else:
    df_top15 = df.nlargest(N, 'deaths')
if sorting:
    df_top15.sort_values(by='deaths', ascending=False, inplace=True)


#df_top15.head()
```

```python

# Create bar plot with matplotlib
plt.figure(figsize=(12, 8))
if ratio:
    percent = '%'
    values = df_top15[target]/df_top15['cases']*100
else:
    values = df_top15[target]
    percent = ''
if values.max() > 100:
    format_str = '{:,.0f}'
else:
    format_str = '{:,.2f}%'
offset = max(values)*0.01

#bars = plt.bar(df_top15['country'], values, color='magenta', alpha=0.7, edgecolor='black')
bars = plt.bar(df_top15['country'], values, 
               color=df_top15['color'], edgecolor='black', linewidth=0.5)
plt.title(f'COVID-19 {target}{percent} in Top {n} Countries', fontsize=16, fontweight='bold')
plt.xlabel('Country', fontsize=12)
plt.ylabel(f'Number of {target}', fontsize=12)
plt.xticks(rotation=90, ha='center')
plt.grid(axis='y', alpha=0.3)

# Add value labels on bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + offset,
             format_str.format(height), ha='center', va='bottom', fontsize=9, rotation=90)

# Add legend
from matplotlib.patches import Patch
legend_elements = [Patch(facecolor=color, label=continent, edgecolor='black') 
                   for continent, color in continent_colors.items() if continent in df_top15['continent'].values]
plt.legend(handles=legend_elements, title='Continents', bbox_to_anchor=(.4, 1), loc='upper left')

plt.tight_layout()
plt.show()

df.to_csv('covid_stats.csv')
```
