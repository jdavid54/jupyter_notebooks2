```python
import random
from collections import deque

class Cell:
    def __init__(self, id, size):
        self.id = id
        self.size = size
        self.up = True
        self.down = True
        self.left = True
        self.right = True

    def neighbors(self):
        row = (self.id - 1) // self.size
        col = (self.id - 1) % self.size
        neigh = {}
        if row > 0: neigh['up'] = self.id - self.size
        if row < self.size - 1: neigh['down'] = self.id + self.size
        if col > 0: neigh['left'] = self.id - 1
        if col < self.size - 1: neigh['right'] = self.id + 1
        return neigh

    def free_sides(self):
        return sum([self.up, self.down, self.left, self.right])

    def no_way_out(self):
        return self.free_sides() < 2

class Maze:
    def __init__(self, size):
        self.size = size
        self.cells = {i: Cell(i, size) for i in range(1, size * size + 1)}
        self._set_boundaries()

    def _set_boundaries(self):
        for i in range(1, self.size + 1):
            self.cells[i].up = False
        for i in range(self.size * (self.size - 1) + 1, self.size * self.size + 1):
            self.cells[i].down = False
        for i in range(1, self.size * self.size + 1, self.size):
            self.cells[i].left = False
        for i in range(self.size, self.size * self.size + 1, self.size):
            self.cells[i].right = False

    def add_wall(self, c1, c2):
        if c2 in self.cells[c1].neighbors().values():
            if self.cells[c1].neighbors().get('up') == c2:
                self.cells[c1].up = False
                self.cells[c2].down = False
            elif self.cells[c1].neighbors().get('down') == c2:
                self.cells[c1].down = False
                self.cells[c2].up = False
            elif self.cells[c1].neighbors().get('left') == c2:
                self.cells[c1].left = False
                self.cells[c2].right = False
            elif self.cells[c1].neighbors().get('right') == c2:
                self.cells[c1].right = False
                self.cells[c2].left = False

    def add_walls_from_list(self, wall_list):
        for w in wall_list:
            self.add_wall(*w)

    def add_random_walls(self, count):
        for _ in range(count):
            c1 = random.randint(1, self.size * self.size)
            neighs = list(self.cells[c1].neighbors().values())
            if neighs:
                c2 = random.choice(neighs)
                self.add_wall(c1, c2)

    def update_states(self):
        return {cid: cell.no_way_out() for cid, cell in self.cells.items()}

    def shortest_path(self, start, end):
        queue = deque([(start, [start])])
        visited = set()
        while queue:
            current, path = queue.popleft()
            if current == end:
                return path
            visited.add(current)
            for dir_name, neigh_id in self.cells[current].neighbors().items():
                if getattr(self.cells[current], dir_name):
                    if neigh_id not in visited:
                        queue.append((neigh_id, path + [neigh_id]))
        return None

    def draw(self, path=None):
        """Draw ASCII maze, optionally highlighting path."""
        path = set(path or [])
        size = self.size

        for r in range(size):
            # Top walls
            line1 = ""
            for c in range(size):
                cell_id = r * size + c + 1
                line1 += "+"
                line1 += "---" if not self.cells[cell_id].up else "   "
            line1 += "+"
            print(line1)

            # Middle (left/right walls)
            line2 = ""
            for c in range(size):
                cell_id = r * size + c + 1
                line2 += "|" if not self.cells[cell_id].left else " "
                if cell_id in path:
                    line2 += " * "  # path marker
                else:
                    line2 += "   "
            line2 += "|" if not self.cells[size * r + size].right else " "
            print(line2)

        # Bottom wall
        print("+" + "---+" * size)
```

```python
import random
from collections import deque

class Cell:
    def __init__(self, cell_id, size):
        self.id = cell_id
        self.size = size
        # True = open side, False = wall
        self.up = True
        self.down = True
        self.left = True
        self.right = True

    def free_sides(self):
        return sum([self.up, self.down, self.left, self.right])

    def no_way_out(self):
        return self.free_sides() < 2

    def neighbors(self):
        nbs = {}
        r = (self.id - 1) // self.size
        c = (self.id - 1) % self.size
        if self.up and r > 0:
            nbs['up'] = self.id - self.size
        if self.down and r < self.size - 1:
            nbs['down'] = self.id + self.size
        if self.left and c > 0:
            nbs['left'] = self.id - 1
        if self.right and c < self.size - 1:
            nbs['right'] = self.id + 1
        return nbs

class Maze:
    def __init__(self, size):
        self.size = size
        self.cells = {i: Cell(i, size) for i in range(1, size * size + 1)}
        self.init_boundaries()

    def init_boundaries(self):
        size = self.size
        # Top row
        for i in range(1, size + 1):
            self.cells[i].up = False
        # Bottom row
        for i in range(size * (size - 1) + 1, size * size + 1):
            self.cells[i].down = False
        # Left column
        for i in range(1, size * size + 1, size):
            self.cells[i].left = False
        # Right column
        for i in range(size, size * size + 1, size):
            self.cells[i].right = False

        # Entry and Exit: exactly 2 sides open for cell 1 and cell N
        self.cells[1].left = True     # Entry side open
        self.cells[1].right = True    # Open right side too

        last = size * size
        self.cells[last].right = True  # Exit side open
        self.cells[last].left = True   # Open left side too


    def add_wall_between(self, a, b):
        # Remove wall between cells a and b if valid
        if abs(a - b) == 1:  # horizontal neighbors
            if a < b:
                self.cells[a].right = False
                self.cells[b].left = False
            else:
                self.cells[b].right = False
                self.cells[a].left = False
        elif abs(a - b) == self.size:  # vertical neighbors
            if a < b:
                self.cells[a].down = False
                self.cells[b].up = False
            else:
                self.cells[b].down = False
                self.cells[a].up = False

    def add_walls_from_list(self, wall_list):
        for a, b in wall_list:
            self.add_wall_between(a, b)

    def add_random_walls(self, count):
        added = 0
        all_pairs = []
        for i in range(1, self.size * self.size + 1):
            for nb in self.cells[i].neighbors().values():
                if i < nb:
                    all_pairs.append((i, nb))
        random.shuffle(all_pairs)

        for a, b in all_pairs:
            # Prevent completely closing off a cell
            if self.cells[a].free_sides() > 1 and self.cells[b].free_sides() > 1:
                self.add_wall_between(a, b)
                added += 1
                if added >= count:
                    break

    def update_states(self):
        """Iterative dead-end detection."""
        deadends = {cid for cid, cell in self.cells.items() if cell.no_way_out()}
        changed = True

        while changed:
            changed = False
            for cid in list(deadends):
                for neigh_id in self.cells[cid].neighbors().values():
                    if neigh_id not in deadends:
                        if self.cells[neigh_id].free_sides() == 2:
                            deadends.add(neigh_id)
                            changed = True

        return {cid: (cid in deadends) for cid in self.cells}

    def shortest_path(self, start, goal):
        q = deque([[start]])
        visited = {start}
        while q:
            path = q.popleft()
            cell_id = path[-1]
            if cell_id == goal:
                return path
            for nb in self.cells[cell_id].neighbors().values():
                if nb not in visited:
                    visited.add(nb)
                    q.append(path + [nb])
        return []

    def draw(self, path=None):
        """Draw maze with optional path (*)"""
        path = set(path or [])
        size = self.size
        for r in range(size):
            line1 = ""
            for c in range(size):
                cell_id = r * size + c + 1
                line1 += "+"
                line1 += "---" if not self.cells[cell_id].up else "   "
            line1 += "+"
            print(line1)

            line2 = ""
            for c in range(size):
                cell_id = r * size + c + 1
                line2 += "|" if not self.cells[cell_id].left else " "
                if cell_id in path:
                    line2 += " * "
                else:
                    line2 += "   "
            line2 += "|" if not self.cells[size * r + size].right else " "
            print(line2)

        print("+" + "---+" * size)

    def draw_with_deadends(self, deadends):
        """Draw maze highlighting dead ends with X"""
        deadends = set(deadends)
        size = self.size
        for r in range(size):
            line1 = ""
            for c in range(size):
                cell_id = r * size + c + 1
                line1 += "+"
                line1 += "---" if not self.cells[cell_id].up else "   "
            line1 += "+"
            print(line1)

            line2 = ""
            for c in range(size):
                cell_id = r * size + c + 1
                line2 += "|" if not self.cells[cell_id].left else " "
                if cell_id in deadends:
                    line2 += " X "
                else:
                    line2 += "   "
            line2 += "|" if not self.cells[size * r + size].right else " "
            print(line2)

        print("+" + "---+" * size)


if __name__ == "__main__":
    size = 10
    maze = Maze(size)
    # Example manual walls
    #maze.add_walls_from_list([(1, 2), (5, 15)])
    # Add minimum 80 walls randomly
    maze.add_random_walls(50)

    states = maze.update_states()
    deadends = [cid for cid, val in states.items() if val]

    path = maze.shortest_path(1, size * size)

    print("\nMaze with shortest path:")
    maze.draw(path)

    print("\nMaze with dead ends marked X:")
    maze.draw_with_deadends(deadends)
```

```python
import random
from collections import deque

class Cell:
    def __init__(self, cell_id, size):
        self.id = cell_id
        self.size = size
        self.up = True
        self.down = True
        self.left = True
        self.right = True
        self.dead_end = False
        self.group_id = None

    def open_sides(self):
        return sum([self.up, self.down, self.left, self.right])

class MazeAI:
    def __init__(self, size):
        self.size = size
        self.cells = {}
        for i in range(1, size*size +1):
            self.cells[i] = Cell(i, size)
        self._init_boundaries()

    def _init_boundaries(self):
        for i in range(1, self.size + 1):  # Top row
            self.cells[i].up = False
        for i in range(self.size*(self.size-1)+1, self.size*self.size + 1):  # Bottom row
            self.cells[i].down = False
        for r in range(self.size):
            self.cells[1 + r*self.size].left = False  # Left col
            self.cells[self.size + r*self.size].right = False  # Right col

        # Entry/Exit
        self.cells[1].left = True
        self.cells[self.size*self.size].right = True

    def add_wall(self, cell_a, cell_b):
        # Add wall between adjacent cells
        diff = cell_b - cell_a
        if abs(diff) > self.size:  # Not adjacent vertically or horizontally
            return
        if diff == 1:
            self.cells[cell_a].right = False
            self.cells[cell_b].left = False
        elif diff == -1:
            self.cells[cell_a].left = False
            self.cells[cell_b].right = False
        elif diff == self.size:
            self.cells[cell_a].down = False
            self.cells[cell_b].up = False
        elif diff == -self.size:
            self.cells[cell_a].up = False
            self.cells[cell_b].down = False

    def get_neighbors(self, cell_id):
        neighbors = []
        c = self.cells[cell_id]
        row = (cell_id -1) // self.size
        col = (cell_id -1) % self.size
        # up
        if row > 0:
            neighbors.append(cell_id - self.size)
        # down
        if row < self.size -1:
            neighbors.append(cell_id + self.size)
        # left
        if col > 0:
            neighbors.append(cell_id -1)
        # right
        if col < self.size -1:
            neighbors.append(cell_id +1)
        return neighbors

    def is_path_open(self, cell_a_id, cell_b_id):
        diff = cell_b_id - cell_a_id
        if diff == 1:  # right neighbor
            return self.cells[cell_a_id].right and self.cells[cell_b_id].left
        elif diff == -1:  # left neighbor
            return self.cells[cell_a_id].left and self.cells[cell_b_id].right
        elif diff == self.size:  # down neighbor
            return self.cells[cell_a_id].down and self.cells[cell_b_id].up
        elif diff == -self.size:  # up neighbor
            return self.cells[cell_a_id].up and self.cells[cell_b_id].down
        return False

    def propagate_dead_ends(self):
        changed = True
        while changed:
            changed = False
            for cell in self.cells.values():
                if cell.dead_end:
                    continue
                open_paths = 0
                neighbors = self.get_neighbors(cell.id)
                for n_id in neighbors:
                    if self.is_path_open(cell.id, n_id):
                        if not self.cells[n_id].dead_end:
                            open_paths += 1
                if open_paths <= 1:
                    cell.dead_end = True
                    changed = True

    def mark_single_access_groups(self):
        visited = set()
        group_counter = 1
        for cell_id, cell in self.cells.items():
            if cell.dead_end or cell_id in visited:
                continue
            group_cells = set()
            queue = [cell_id]
            while queue:
                c_id = queue.pop()
                if c_id in group_cells:
                    continue
                c = self.cells[c_id]
                if c.dead_end:
                    continue
                group_cells.add(c_id)
                neighbors = self.get_neighbors(c_id)
                for n_id in neighbors:
                    if n_id not in group_cells and not self.cells[n_id].dead_end:
                        queue.append(n_id)
            visited.update(group_cells)
            external_accesses = set()
            for gc_id in group_cells:
                neighbors = self.get_neighbors(gc_id)
                for n_id in neighbors:
                    if n_id not in group_cells and self.is_path_open(gc_id, n_id):
                        external_accesses.add(n_id)
            if len(external_accesses) == 1:
                gid = f'G{group_counter}'
                group_counter += 1
                for gc_id in group_cells:
                    self.cells[gc_id].group_id = gid

    def get_maze_status(self):
        status = {}
        for cell_id, cell in self.cells.items():
            status[cell_id] = {
                "dead_end": cell.dead_end,
                "group_id": cell.group_id,
                "walls": {
                    "up": cell.up,
                    "down": cell.down,
                    "left": cell.left,
                    "right": cell.right,
                }
            }
        return status

    def print_maze(self):
        size = self.size
        for r in range(size):
            # Print top walls
            line = ""
            for c in range(size):
                cell = self.cells[r*size + c +1]
                line += "+" + ("   " if cell.up else "---")
            line += "+"
            print(line)
            # Print left walls and cell marks
            line = ""
            for c in range(size):
                cell = self.cells[r*size + c +1]
                left_wall = "|" if not cell.left else " "
                # Mark: dead_end = 'X', group_id = G#, else ' '
                if cell.dead_end:
                    mark = "X"
                elif cell.group_id is not None:
                    mark = cell.group_id[-1]  # Just number for brevity
                else:
                    mark = " "
                line += f"{left_wall} {mark} "
            # right wall of last cell
            last_cell = self.cells[r*size + size]
            line += "|" if not last_cell.right else " "
            print(line)
        # Bottom walls
        line = ""
        for c in range(size):
            cell = self.cells[size*size - size + c +1]
            line += "+" + ("---" if not cell.down else "   ")
        line += "+"
        print(line)

# Example usage:
if __name__ == "__main__":
    size = 10
    maze = MazeAI(size)

    # Example: add some walls here (or use your own wall list)
    walls_to_add = [
        # Example walls for testing — add your own walls here
        (5, 6), (6, 7), (7, 8), (8, 9),
        (15, 16), (16, 17), (17, 18), (18, 19),
        (25, 26), (26, 27), (27, 28), (28, 29),
        (37, 38), (38, 39)
    ]
    for w in walls_to_add:
        maze.add_wall(*w)

    maze.propagate_dead_ends()
    maze.mark_single_access_groups()
    maze.print_maze()

    import json
    status = maze.get_maze_status()
    print(json.dumps(status, indent=2))
```

# Trees of maze


```python
import random
from collections import deque

class CellAgent:
    def __init__(self, cell_id, size):
        self.id = cell_id
        self.size = size
        self.up = True
        self.down = True
        self.left = True
        self.right = True
        self.dead_end = False
        self.group_id = None  # For group marking

    def free_sides(self):
        return sum([self.up, self.down, self.left, self.right])

    def get_neighbor_ids(self):
        nbs = {}
        r = (self.id - 1) // self.size
        c = (self.id - 1) % self.size
        if self.up and r > 0:
            nbs['up'] = self.id - self.size
        if self.down and r < self.size - 1:
            nbs['down'] = self.id + self.size
        if self.left and c > 0:
            nbs['left'] = self.id - 1
        if self.right and c < self.size - 1:
            nbs['right'] = self.id + 1
        return nbs

class MazeAI:
    def __init__(self, size):
        self.size = size
        self.cells = {i: CellAgent(i, size) for i in range(1, size*size + 1)}
        self.init_boundaries()

    def init_boundaries(self):
        size = self.size
        for i in range(1, size + 1):
            self.cells[i].up = False
        for i in range(size*(size-1) + 1, size*size + 1):
            self.cells[i].down = False
        for i in range(1, size*size + 1, size):
            self.cells[i].left = False
        for i in range(size, size*size + 1, size):
            self.cells[i].right = False

        self.cells[1].left = True    # Entry open side
        self.cells[size*size].right = True  # Exit open side

    def add_wall_between(self, a, b):
        if abs(a - b) == 1:
            if a < b:
                self.cells[a].right = False
                self.cells[b].left = False
            else:
                self.cells[b].right = False
                self.cells[a].left = False
        elif abs(a - b) == self.size:
            if a < b:
                self.cells[a].down = False
                self.cells[b].up = False
            else:
                self.cells[b].down = False
                self.cells[a].up = False

    def add_random_walls(self, count):
        added = 0
        all_pairs = []
        for i in range(1, self.size*self.size + 1):
            for nb in self.cells[i].get_neighbor_ids().values():
                if i < nb:
                    all_pairs.append((i, nb))
        random.shuffle(all_pairs)

        for a, b in all_pairs:
            if self.cells[a].free_sides() > 1 and self.cells[b].free_sides() > 1:
                self.add_wall_between(a, b)
                added += 1
                if added >= count:
                    break

    def build_bfs_tree(self, start=1):
        visited = set([start])
        tree = {}
        parent_map = {start: None}
        queue = deque([start])

        while queue:
            node = queue.popleft()
            tree[node] = []
            for nbr in self.cells[node].get_neighbor_ids().values():
                if nbr not in visited:
                    visited.add(nbr)
                    tree[node].append(nbr)
                    parent_map[nbr] = node
                    queue.append(nbr)
        return tree, parent_map

    def mark_dead_ends_and_groups(self, tree):
        # Dead ends = leaves in BFS tree
        dead_end = set(node for node, children in tree.items() if len(children) == 0)

        # Propagate dead end status upwards if all children dead ends
        changed = True
        while changed:
            changed = False
            for node, children in tree.items():
                if node in dead_end:
                    continue
                if children and all(child in dead_end for child in children):
                    dead_end.add(node)
                    changed = True

        # Find connected groups of dead ends
        visited = set()
        groups = []

        def dfs_group(node, group_nodes):
            group_nodes.append(node)
            visited.add(node)
            for c in tree.get(node, []):
                if c in dead_end and c not in visited:
                    dfs_group(c, group_nodes)

        for node in dead_end:
            if node not in visited:
                group_nodes = []
                dfs_group(node, group_nodes)
                groups.append(group_nodes)

        # For each group, count access edges (edges to outside group)
        def count_access(group):
            group_set = set(group)
            access = 0
            for cell_id in group:
                neighbors = self.cells[cell_id].get_neighbor_ids().values()
                for nb in neighbors:
                    if nb not in group_set:
                        access += 1
            return access

        # Mark dead ends and groups in cells
        for cell in self.cells.values():
            cell.dead_end = (cell.id in dead_end)
            cell.group_id = None

        self.single_access_groups = []
        gid = 1
        for group in groups:
            if count_access(group) == 1:
                self.single_access_groups.append(set(group))
                for cell_id in group:
                    self.cells[cell_id].group_id = gid
                gid += 1

    def find_shortest_path(self, start=1, goal=None):
        if goal is None:
            goal = self.size * self.size
        queue = deque([start])
        visited = {start}
        parent = {start: None}

        while queue:
            node = queue.popleft()
            if node == goal:
                path = []
                while node is not None:
                    path.append(node)
                    node = parent[node]
                return path[::-1]
            for nbr in self.cells[node].get_neighbor_ids().values():
                if nbr not in visited:
                    # Avoid dead ends in path
                    if not self.cells[nbr].dead_end:
                        visited.add(nbr)
                        parent[nbr] = node
                        queue.append(nbr)
        return None

    def draw(self, path=None):
        size = self.size
        path_set = set(path) if path else set()

        for r in range(size):
            # Top walls
            line1 = ""
            for c in range(size):
                cell_id = r*size + c + 1
                line1 += "+"
                line1 += "---" if not self.cells[cell_id].up else "   "
            line1 += "+"
            print(line1)

            # Left walls and cell content
            line2 = ""
            for c in range(size):
                cell_id = r*size + c + 1
                line2 += "|" if not self.cells[cell_id].left else " "

                if cell_id in path_set:
                    line2 += " * "
                elif self.cells[cell_id].dead_end:
                    line2 += " X "
                elif self.cells[cell_id].group_id is not None:
                    line2 += " G "
                else:
                    line2 += "   "
            line2 += "|" if not self.cells[size*r + size].right else " "
            print(line2)
        print("+" + "---+"*size)

```
