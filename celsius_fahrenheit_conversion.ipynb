# Celsius and Fahrenheit conversion

Use the reusable `temperature_conversion` module to convert temperatures.

```python
from temperature_conversion import celsius_to_fahrenheit, fahrenheit_to_celsius
```

```python
celsius = 25
fahrenheit = celsius_to_fahrenheit(celsius)
print(f"{celsius}°C = {fahrenheit}°F")
```

```python
fahrenheit = 77
celsius = fahrenheit_to_celsius(fahrenheit)
print(f"{fahrenheit}°F = {celsius}°C")
```

```python
for t in range(-20, 51, 10):
    fahrenheit = celsius_to_fahrenheit(t)
    print(f"{t}°C = {fahrenheit}°F")    
```
