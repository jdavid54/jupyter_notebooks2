```python
# subscriber.py
import paho.mqtt.client as mqtt
# https://eclipse.dev/paho/files/paho.mqtt.python/html/migrations.html
import paho.mqtt
version = paho.mqtt.__version__
# importing whole module
# from tkinter import *
# from tkinter.ttk import *
# from tkinter_rainbow import *
import random
print(version)
```

```python
# create connection, the three parameters are broker address, broker port number, and keep-alive time respectively
brokers = ["test.mosquitto.org","broker.emqx.io","192.168.0.175"]
broker_id = 2 
broker = brokers[broker_id]
client_id = f'python-mqtt-{random.randint(0, 1000)}'
print(client_id)
```

```python
if version != '2.1.0':
    client = mqtt.Client(client_id)
        
else:    # paho mqtt version 2
    client = mqtt.Client(client_id=client_id, callback_api_version=mqtt.CallbackAPIVersion.VERSION2)
```

```python
import time
import threading
# Save timestamp
start = time.time()
last_msg_time = start
INACTIVITY_TIMEOUT = 100 # délai d'attente entre 2 msg
```

```python
if version != '2.1.0':
    def on_connect(client, userdata, flags, rc):  # version 1.6
        print(f"Connected with result code {rc}")
        print("CTRL-Z to stop")
        # subscribe, which need to put into on_connect
        # if reconnect after losing the connection with the broker, it will continue to subscribe to the raspberry/topic topic
        client.subscribe("conso_day")
        # paho mqtt version 1.6.1    
else:       
    def on_connect(client, userdata, flags, rc, properties):  # version 2.0
        print(f"Connected with result code {rc}")
        print("CTRL-Z to stop")
        # subscribe, which need to put into on_connect
        # if reconnect after losing the connection with the broker, it will continue to subscribe to the raspberry/topic topic
        client.subscribe("conso_day")
   
# the callback function, it will be triggered when receiving messages
def on_message(client, userdata, msg):
    global start, last_msg_time
    print(f"{msg.topic} {msg.payload}")
    # Save timestamp
    end = time.time()
    t = end - start
    start = end
    print('intervalle:',t)
    #print(f'intervalle: {t}')
    last_msg_time = end
    print(time.strftime("%Y-%m-%d %H:%M:%S", time.localtime(last_msg_time)), f"{round(t,2)}s")
    try:
        kw = int(msg.payload)/1000
        value = str(kw)+' kW'
        print(value)
        #lbl.config(text = value)
        #change_color(int(msg.payload)//10)        
        #lbl2.config(text = f"{round(t,2)}s")      
    except:
        pass

```

```python
client.on_connect = on_connect
client.on_message = on_message
#client.on_disconnect = on_disconnect
client.connect(broker, 1883, 60)  #broker local
client.loop_start()
```
