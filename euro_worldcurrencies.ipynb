```python
pdframe['Dernier'] = pd.to_numeric(pdframe['Dernier'], errors='coerce')
pdframe_sorted = pdframe.sort_values('Dernier', ascending=True)
pdframe_sorted = pdframe_sorted.reset_index(drop=True)
pdframe_sorted.head(10)
```

```python
pdframe_sorted.tail(20)
```

```python
import matplotlib.pyplot as plt

pdframe_sorted.tail(50).plot(x='Libellé', y='Dernier', figsize=(12, 6), logy=True,
                             title='Currency Exchange Rates (Sorted)', kind='bar', legend=False)
plt.xticks(rotation=90, ha='right', fontsize=8)
plt.tight_layout()
plt.show()
```
