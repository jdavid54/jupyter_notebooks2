```python
pdframe['Dernier'] = pd.to_numeric(pdframe['Dernier'], errors='coerce')
pdframe_sorted = pdframe.sort_values('Dernier', ascending=True)
pdframe_sorted = pdframe_sorted.reset_index(drop=True)
pdframe_sorted.head(10)
```

```python
pdframe_sorted.tail(20)
```

```python
import matplotlib.pyplot as plt
n = 20
pdframe_sorted.tail(n).plot(x='Libellé', y='Dernier', figsize=(12, 6), logy=True,
                             title='Currency Exchange Rates (Sorted)', kind='bar', legend=False)
plt.xticks(rotation=90, ha='right', fontsize=8)
plt.tight_layout()
plt.show()
```

```python
subset = pdframe_sorted.tail(n)

plt.figure(figsize=(12, 6))
plt.barh(subset['Libellé'], subset['Dernier'], color='steelblue')
plt.xlabel('Dernier')
plt.ylabel('Libellé')
plt.title('Currency Exchange Rates (Sorted)')
plt.tight_layout()
plt.show()
```
