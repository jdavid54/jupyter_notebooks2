```python
import json
import pandas as pd

file_path = r"Y:\Documents\horaires_RER_E.json"

with open(file_path, "r", encoding="utf-8") as file:
    data = json.load(file)

# Find "MonitoredStopVisit" even if it is nested in the JSON structure.
def find_key(obj, key):
    if isinstance(obj, dict):
        if key in obj:           
            return obj[key]
        for value in obj.values():
            result = find_key(value, key)
            if result is not None:
                return result
    elif isinstance(obj, list):
        for item in obj:
            result = find_key(item, key)
            if result is not None:
                return result
            return result
    return None

visits = find_key(data, "MonitoredStopVisit")
```

```python
if visits is None:
    raise KeyError('"MonitoredStopVisit" was not found in the JSON file.')

if isinstance(visits, dict):
    visits = [visits]

df = pd.json_normalize(visits, sep=".")
# Convert columns whose names indicate dates or timestamps.
for column in df.columns:
    if "date" in column.lower() or "time" in column.lower():
        df[column] = pd.to_datetime(df[column], errors="coerce")

df.head()
```
