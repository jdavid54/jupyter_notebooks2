Pour intégrer un fichier `.md` dans un notebook Jupyter, plusieurs solutions sont possibles.

### Créer un lien vers le fichier

```markdown
[Ouvrir le fichier Markdown](chemin/vers/fichier.md)
```

Exemple :

```markdown
[Documentation](./documentation.md)
```

### Afficher son contenu dans le notebook

Dans une cellule **Python** :

```python
from pathlib import Path
from IPython.display import Markdown, display

contenu = Path("documentation.md").read_text(encoding="utf-8")
display(Markdown(contenu))
```

Si le fichier est dans un sous-dossier :

```python
contenu = Path("docs/documentation.md").read_text(encoding="utf-8")
display(Markdown(contenu))
```

Le Markdown ne possède pas de commande native permettant d’inclure directement le contenu d’un autre fichier.
